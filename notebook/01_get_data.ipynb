# 01 ดึงข้อมูล
อ่านไฟล์ `Clean_BangPaKong.xlsx` (กรมชลประทาน ฉะเชิงเทรา) แล้วบันทึกสำเนาข้อมูลดิบไว้ใน `data/raw` พร้อมวันที่ดึงข้อมูล

```python
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
RAW = ROOT / "data" / "raw"
PROC = ROOT / "data" / "processed"
FIG = ROOT / "figures" / "regenerated"   # รูปที่ notebook สร้างใหม่ (รูปในเล่มอยู่ที่ figures/)
RES = ROOT / "results"
for p in (RAW, PROC, FIG, RES):
    p.mkdir(parents=True, exist_ok=True)
```

```python
from datetime import date
import pandas as pd

file_id = "17Rhdc_B2VhpVXhngCWWYakU8LTq41nL"
url = f"https://drive.google.com/uc?export=download&id={file_id}"
local = RAW / "Clean_BangPaKong.xlsx"          # ใช้ไฟล์ในเครื่องก่อน ถ้ามี

df = pd.read_excel(local if local.exists() else url)
out = RAW / f"BangPaKong_raw_{date.today():%Y-%m-%d}.csv"
df.to_csv(out, index=False)
print(df.shape, "->", out.name)
df.head()
```
