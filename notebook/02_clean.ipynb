# 02 เตรียมข้อมูล
ตรวจค่าสูญหายและแถวซ้ำ ลบแถวซ้ำ แปลงชนิดข้อมูล และสร้าง Year, Month

```python
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
RAW = ROOT / "data" / "raw"
PROC = ROOT / "data" / "processed"
FIG = ROOT / "figures" / "regenerated"   # รูปที่ notebook สร้างใหม่ (รูปในเล่มอยู่ที่ figures/)
RES = ROOT / "results"
for p in (RAW, PROC, FIG, RES):
    p.mkdir(parents=True, exist_ok=True)
```

```python
import pandas as pd

raw_file = sorted(RAW.glob("BangPaKong_raw_*.csv"))[-1]   # ไฟล์ดิบล่าสุด
df = pd.read_csv(raw_file)
print("ไฟล์ดิบ:", raw_file.name, df.shape)
df.info()
print(df.isna().sum())
print("แถวซ้ำทั้งแถว:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()                       # เก็บแถวแรกของแต่ละกลุ่มที่ซ้ำ
df["DateTime"] = pd.to_datetime(df["DateTime"])
df = df.rename(columns={"WaterLevel(m)": "WaterLevel"})
df = df.sort_values("DateTime").reset_index(drop=True)
df["Year"] = df["DateTime"].dt.year
df["Month"] = df["DateTime"].dt.month

print("n หลังลบแถวซ้ำ =", f"{len(df):,}", "(ในเล่มคือ 352,684)")
df.to_csv(PROC / "BangPaKong_clean.csv", index=False)
df.head()
```
